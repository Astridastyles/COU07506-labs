```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from scipy.io import wavfile

plt.rcParams["figure.figsize"] = (9, 3.5)
```

```python
# UNIT IMPULSE SIGNAL

n = np.arange(-5, 6)

delta = (n == 0).astype(int)

plt.stem(n, delta)

plt.title("Unit Impulse Signal")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
# UNIT STEP SIGNAL

n = np.arange(-5, 11)

step = (n >= 0).astype(float)

plt.stem(n, step)

plt.title("Unit Step Signal")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
# UNIT RAMP SIGNAL

n = np.arange(-5, 11)

step = (n >= 0).astype(float)

ramp = n * step

plt.stem(n, ramp)

plt.title("Unit Ramp Signal")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
# REAL EXPONENTIAL SIGNAL

n = np.arange(-5, 11)

step = (n >= 0).astype(float)

expo = 0.8**n * step

plt.stem(n, expo)

plt.title("Real Exponential Signal")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()

plt.show()
```

```python
# TASK 1.1 - ALL FOUR SIGNALS

n = np.arange(-5, 11)

delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["δ(n)", "u(n)", "r(n)", "(0.8)^n u(n)"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")

plt.tight_layout()
plt.show()
```

Task 1.2

```python
# TASK 1.2 - EXPONENTIAL SIGNALS

n = np.arange(-5, 11)

step = (n >= 0).astype(float)

a_values = [0.5, -0.8, 1.1]

fig, axes = plt.subplots(1, 3, figsize=(13, 3))

for ax, a in zip(axes, a_values):

    x = (a**n) * step

    ax.stem(n, x)

    ax.set_title(f"a = {a}")
    ax.set_xlabel("n")
    ax.set_ylabel("Amplitude")
    ax.grid()

plt.tight_layout()
plt.show()
```

```python
# TASK 1.2 - COSINE SIGNALS

n = np.arange(0, 41)

x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

plt.figure(figsize=(12, 4))

plt.stem(
    n, x1,
    linefmt='b-',
    markerfmt='bo',
    basefmt='k-',
    label='Cosine period 8'
)

plt.stem(
    n, x2,
    linefmt='r-',
    markerfmt='ro',
    basefmt=' ',
    label='Cosine period 7.5'
)

plt.title("Comparison of Two Cosine Signals")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()

plt.show()
```

Observations:
When a = 0.5, the exponential signal decreases gradually and remains positive
When a = -0.8 the signal alternates between positive and negative values while its amplitude decreases
When a = 1.1, the exponential signal increases with time
The cosine signal with period 8 repeats every 8 samples while the cosine signal with period 7.5 repeats every 15 samples.
Conclusion: Both cosine signals are periodic but they have different periods
